## Data description
The data available are images that are in two formarts. One is [dicom](https://www.dicomstandard.org/), the recommended formart for medical images, whilst the other is in PNG. The images are chest x-ray.

### Positive images
These were sourced from the NIH portal and are in the formart of dicom. These have an id in the formart of [uuid](https://en.wikipedia.org/wiki/Universally_unique_identifier). And are double nested.

Initially when downoload, the files are in zip formart and upon extraction usingthey are in this is how the file structure looks like.

```
650761c9-79fd-43e9-a50b-ffd4bcd14039                                              ← Patient ID (UUID)
└── 1.2.826.0.1.3680043.2.1125.1.1987707188258529255253844547861857               ← Study Instance UID
    └── 1.2.826.0.1.3680043.2.1125.1.84803483488010727226141612467471751          ← Series Instance UID
        └── 1.2.826.0.1.3680043.2.1125.1.50634073513295032690663563825745853.dcm  ← SOP Instance UID (the actual file)
```

##### Patient ID(UUID)
This is a patient Id that has been annonymized already.

##### Study Instance UID
  This is the key that is used directly, in the NIH portal.
##### Series instance UID
  These represnet imagages that were grouped collected together under same technical conditions during one study.
##### Service Object Pair (SOP UID)
This is the leaf of the hierachy and it is the id of the dicom file.
